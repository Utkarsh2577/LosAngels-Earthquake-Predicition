```python
import matplotlib.pyplot as plt
import seaborn as sns
import math
```

```python
# Calculating and plotting the correlation matrix between X and Y
correlation_matrix = df.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix of Features and Target")

# Save the plot as an image file
plt.savefig("correlation_matrix_heatmap.png", dpi=300, bbox_inches='tight')

# Display the plot
plt.show()
```

```python
num_cols = len(X.columns)
num_rows = math.ceil(num_cols / 2)

fig, axes = plt.subplots(num_rows, 2, figsize=(16, num_rows * 6))
axes = axes.flatten()

for i, column in enumerate(X.columns):
    sns.histplot(X[column], kde=True, ax=axes[i]) 
    axes[i].set_title(f'Distribution of {column}')
    axes[i].set_xlabel(column)
    axes[i].set_ylabel('Frequency')
if num_cols % 2 != 0:
    axes[-1].set_visible(False)

plt.tight_layout()
plt.savefig("feature_distributions.png", dpi=300, bbox_inches='tight')
plt.show()

```

```python
from sklearn.model_selection import train_test_split
```

```python
# Splitting the dataset into train and test with 70% and 30 %
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
# Training Random Forest Classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_train, Y_train)
Y_pred_rf = rf_classifier.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Evaluating Random Forest Classifier
print("** Random Forest Classifier Results **")
print(f"Accuracy: {accuracy_score(Y_test, Y_pred_rf)}")
print("Classification Report:")
print(classification_report(Y_test, Y_pred_rf))

```

```python
import xgboost as xgb
```

```python
# Adjusting class labels to start from 0
Y_train_adjusted = Y_train - 1
Y_test_adjusted = Y_test - 1

# Training XGBoost Classifier
xgb_classifier = xgb.XGBClassifier(objective='multi:softmax', num_class=6, n_estimators=100, random_state=42)
xgb_classifier.fit(X_train, Y_train_adjusted)
Y_pred_xgb = xgb_classifier.predict(X_test)

# Mapping predictions back to original range (1 to 6)
Y_pred_xgb = Y_pred_xgb + 1

# Evaluating XGBoost Classifier
print("\n** XGBoost Classifier Results **")
print(f"Accuracy: {accuracy_score(Y_test, Y_pred_xgb):.4f}")
print("Classification Report:")
print(classification_report(Y_test, Y_pred_xgb))
```
